```python
# add all prime numbers from 1 to n and n is given by the user
n = int(input("enter number: "))
total_sum = 0

for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        total_sum += num

print("Sum of prime numbers from 1 to", n, "is:", total_sum)
```

```python
#assignment 6.1 and 6.2
# add all odd numbers from 1 to n, n is given by user
def sum_of_odds(n):
    total_sum = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total_sum += i
    return total_sum

n = int(input("Enter the number: "))
result = sum_of_odds(n)

print("Sum of odd numbers from 1 to", n, "is:", result)
```

```python
#add all prime numbers from to n, n is given by user
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

n = int(input("Enter the number: "))
total_sum = 0

for i in range(1, n + 1):
    if is_prime(i):
        total_sum += i

print("Sum of prime numbers from 1 to", n, "is:", total_sum)
```
